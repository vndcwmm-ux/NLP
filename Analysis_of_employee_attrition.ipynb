## Methodology

Before modeling and analysis, data was cleaned and prepared. Missing values were handeled, categorial variables were encoded, duplicates were removed, attrition variable was converted to numeric: Yes=1, No=0
Explaratory data analysis (EDA) was performed.
For this descriptive statistics (means, proportions by group) for features vs. attrition were examined — to observe which features differ between leavers and stayers.
Also, visualization (boxplots, countplots, correlation heatmap) was made — to inspect distributions, relationships, possible nonlinearities, and initial hypotheses testing (e.g. differences in income, job satisfaction, travel frequency).
Finally, statistical test as chi-square test for categorical variables (like BusinessTravel, OverTime) vs. Attrition was made.
Predicting modelling was performed by using such models as:
Logistic Regression — a baseline, interpretable model for binary classification. As a linear model, LR allows understanding the direct (linear) association between each predictor and attrition probability[4].
Random Forest — a tree‑based method that can capture non-linear relationships, interactions between features, and is robust to overfitting. RF is frequently found to perform well in attrition‑prediction tasks[1]. And Gradient Boosting which is a sequential ensemble method that optimizes an arbitrary loss function using gradient descent[1].
Model evaluation will be based on standard classification metrics: accuracy, precision, recall, F1‑score and ROC/AUC.

According to a recent systematic review, nearly all studies on employee attrition prediction over the past decade use supervised machine learning on cross‑sectional HR datasets.Combining interpretable models (Logistic regression) with powerful non-linear models (Random forest) balances the trade-off between explainability and predictive performance, which is considered best practice in applied HR analytics

##Machine learning

Our key goal in terms of usage of machine learning is to predict the attrition of employee and analyse what key feuatures affect it.

Firstly, we need to check how the data is distributed among Attrition variable:

```python
df['Attrition'].value_counts()
(df['Attrition'].value_counts(normalize=True)*100).round(4)
```

As can be seen, the proportion of dismsissed employees is about 16%. This means that a significant class imbalance is shown: there are significantly more employees who stayed in the company than those who quit. This is typical for staff turnover forecasting tasks, where layoffs are usually a minority of the total number of employees[2].

The machine learning task within the analysis (predicting staff turnover) is a classification task (supervised learning), therefore, machine learning models were selected based on the task. Moreover, the choice of particular machine learning models  was based on the principles of comparative analysis, interpretability and practical applicability. In our analysis we choose:

* Logistic regression as a base-line model due to interpretability of its results and efficiency with limited data (good balance between bias and variance)
* Random forest due to its resistance to overfitting and processing of nonlinear dependencies
* Gradient boosting as thee most advanced method due to iterative improvement of predictions by minimizing the loss function through gradient descent which leads to high predictive power

To improve the data quality, the following columns were deleted:
*Over18, EmployeeCount, EmployeeNumber, JobLevel, StandardHours, StockOptionLevel, AgeGroup*

These features have been removed for several reasons:
* *Over18, EmployeeCount, StandardHours* are constant values, therefore, have no informational value for the model
* *EmployeeNumber, EmpID* are identifiers, which shoud not affect the forecast
* *JobLevel, StockOptionLevel* are categorical variables encrypted with numbers, information about these variables is not available, so they are not interpretable
* *AgeGroup* duplicates the Age variable in meaning

```python
df = df.drop(['Over18', 'EmployeeCount', 'EmployeeNumber', 'JobLevel', 'StandardHours', 'StockOptionLevel', 'EmpID', 'AgeGroup'], axis=1)
```

Next, we transform the binary variables:
* *OverTime, Gender* have been converted to binary format (0/1)

The remaining categorical features were divided (numeric and categorical) into groups for further processing.

```python
df['OverTime'] = df['OverTime'].map({'Yes': 1, 'No': 0})
df['Gender'] = df['Gender'].map({'Male': 1, 'Female': 0})
```

```python
bin_features = ['Gender', 'OverTime']
cat_features = ['BusinessTravel', 'Department', 'Education', 'EducationField', 'EnvironmentSatisfaction', 'JobInvolvement', 'JobRole', 'JobSatisfaction', 'MaritalStatus',
                'PerformanceRating', 'RelationshipSatisfaction', 'SalarySlab', 'WorkLifeBalance']
num_features = ['Age', 'DailyRate', 'DistanceFromHome', 'HourlyRate', 'MonthlyIncome', 'MonthlyRate', 'NumCompaniesWorked', 'PercentSalaryHike', 'TotalWorkingYears', 'TrainingTimesLastYear',
                'YearsAtCompany', 'YearsInCurrentRole', 'YearsSinceLastPromotion', 'YearsWithCurrManager']
```

The data was divided into **training (80%) and test (20%)** samples while maintaining class proportions (stratify=y). This is important to ensure that the samples are representative, especially when working with unbalanced data.

The ColumnTransformer preprocessor was created, which performs three types of transformations:
* OneHot encoding for categorical features
* Standardization for numerical features
* Transmission without changes for binary features

After the transformation, the number of features, therefore, the dimension of the feature space increased significantly due to OneHot encoding of categorical variables.

```python
X = df[bin_features + cat_features + num_features]
y = df['Attrition']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.2, random_state = 12345, stratify = y)
```

```python
preprocessor = ColumnTransformer(
    transformers=[
        ('cat', OneHotEncoder(), cat_features),
        ('num', StandardScaler(), num_features),
        ('binary', 'passthrough', bin_features)
])
```

```python
X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed =  preprocessor.transform(X_test)
```

```python
feature_names = preprocessor.get_feature_names_out()

X_train_df = pd.DataFrame(X_train_processed, columns = feature_names, index = X_train.index)
X_test_df = pd.DataFrame(X_test_processed, columns = feature_names, index = X_test.index)
```

Now we can check the shape of test and train datasets:

```python
print(f'Train dataset shape: {X_train_df.shape}')
print(f'Test dataset shape: {X_test_df.shape}')
```

Now we can proceed to the modelling. As was stated before, we choose three basic models in our analysis:
* **Logistic regression** as a base-line model due to interpretability of its results and efficiency with limited data (good balance between bias and variance)
* **Random forest due** to its resistance to overfitting and processing of nonlinear dependencies
* **Gradient boosting** as thee most advanced method due to iterative improvement of predictions by minimizing the loss function through gradient descent which leads to high predictive power

```python
models = {
"Logistic Regression":
        LogisticRegression(random_state = 12345,
                           max_iter = 1000,
                           class_weight = 'balanced'),
    "Random Forest":
        RandomForestClassifier(random_state = 12345,
                               n_estimators = 100,
                               class_weight = 'balanced'),
    "Gradient Boosting":
        GradientBoostingClassifier(random_state = 12345,
                                   n_estimators = 100,
                                   learning_rate = 0.1)
}
```

```python
results = []
best_models = {}

for model_name, model in models.items():
  #print(f'\nModel training: {model_name}')
  model.fit(X_train_processed, y_train)

  y_pred = model.predict(X_test_processed)
  y_pred_proba = model.predict_proba(X_test_processed)[:, 1]

  accuracy = accuracy_score(y_test, y_pred)
  precision = precision_score(y_test, y_pred, zero_division=0)
  recall = recall_score(y_test, y_pred)
  f1 = f1_score(y_test, y_pred)
  roc_auc = roc_auc_score(y_test, y_pred_proba)

  results.append({'Model': model_name, 'Accuracy': accuracy, 'Precision': precision,
                  'Recall': recall, 'F1': f1, 'ROC-AUC': roc_auc})

  best_models[model_name] = model
```

To understand predictions of the models we firstly consider confusion matrices for all models. Basically, confusion matrix is a table showing the distribution of predictions of the classification model into four categories:
* True Negative (TN): correctly predicted cases of "employee staying"
* False Positive (FP): the employee is predicted to "quit", but remains
* False Negative (FN): missed dismissals: the employee is predicted to "stay", but quits
* True Positive (TP): correctly predicted attrition

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))
for i, (model_name, model) in enumerate(best_models.items()):
  y_pred = model.predict(X_test_processed)
  cm = confusion_matrix(y_test, y_pred)

  ax = axes[i]
  sns.heatmap(cm, annot=True, fmt='d', cmap='Greens', ax=ax)
  ax.set_title(model_name)
  ax.set_xlabel('Predicted Attrition')
  ax.set_ylabel('Actual Attrition')

plt.suptitle('Condusion matrices', fontsize=14, fontweight='bold')
plt.tight_layout()
plt.show()
```

```python
for model_name, model in best_models.items():
  y_pred = model.predict(X_test_processed)
  cm = confusion_matrix(y_test, y_pred)
  cm_df = pd.DataFrame(cm,
                       index=['Actual: 0', 'Actual: 1'],
                       columns=['Predicted: 0', 'Predicted: 1'])

  print(f"\nModel: {model_name}")
  print(cm_df)
```

As can be seen from confusion matrices, all models accurately predict class 0 (TN), which is expected due to the class imbalance. However, the number of accurately predicted class 1 (TP) significantly varies among the models. Number of FP and FN significantly varies as well. Therefore, we can proceed to the evaluation of quality of the models using metrics calculated based on the confusion matrix.

In order to evaluate quality of models we considered five **key metrics**:
* **Accuracy** shows the percentage of correct predictions from the total number. However, with unbalanced data, which is present, this metric can be misleading, since the model, which always predicts the majority of the class, will have high accuracy.
* **Precision** shows the accuracy of a positive class, which in our analysis is the proportion of predicted attrition that are actually attrition. It is important to minimize false positives.
* **Recall** shows how much of the actual attrition the model was able to detect, which is essential to minimize missed attrition.
* **F1-Score** is the harmonic mean between precision and recall. The optimal metric for unbalanced data.
* **ROC-AUC** is the area under the ROC curve, which shows the model's ability to distinguish classes. A value of 0.5 corresponds to a random classifier, and 1.0 corresponds to a perfect model.

Models' metrics are presented in the table below.

```python
results_df = pd.DataFrame(results)
results_df = results_df.sort_values('F1', ascending = False).reset_index(drop = True)

results_df.round(3)
```

Accordingly to each metrics the best model is:
* Accuracy: Gradient Boosting
* Precision: Random Forest
* Recall: Logistic Regression
* F1 score: Logistic Regression
* ROC_AUC: Logistic Regression

We have **class imbalance**, therefrore, accuracy may be misleading in this case. The metrics **ROC-AUC** is quite representative in our case since it shows the model's ability to distinguish between classes and good for comparing models based on unbalanced data.

Analysing **presicion, recall and F1-score we need to consider particular buisness tasks**, depending on which some metrics are more important.
* Firstly, **precision** shows the percentage of correctly found dismissals, which in terms of buisness means **saving HR resources** (if precision is low, HR spends time on employees who will not leave anyway). This metrics may be the most importnant one in case resources limitation, for instance small HR team.
* Secondly, **recall** shows the percentage of actual attrition detected, which in terms of **buisness means loss prevention** (if recall is low, the company loses employees without warning). This metrics, in turn, may be leading in such case, when key employees are critically important.
* Thirdly, **F1-score** is a harmonic mean that is stable to extreme values and balances precision and recall, which is critical for unbalanced classes. **This metric is optimal** when both FN and FP are important. In terms of buisness this is and optimal metrics.

So, in terms of ROC-AUC as the model's predictive power the results of all model are approximately the same, however, the metrics for Logistic Regression is slightly higher which means that according to ROC-AUC Logistic Regression is the best model. Considering F1-score, Logistic Regression shows the best result as well, which is significantly higher than the rest models' results. Still, depending, on the buisness case, Random Forest is the best model by precision and Logistic Regression -- by Recall. **In general both buisness and academic terms the bost model is Logistic Regression, considering all metrics**.

Overall, analysing features affecting attrition we can split all  top-importtant features into categories:

1. **Economic factors**
The most interesting discovery here is that it is traditionally believed that salary is the main factor in dismissal. Our models show a more complex picture:
* *MonthlyIncome* is really important, both as an absolute value and as a relative value — how much salary meets expectations and market offers
* *OverTime**: employees who regularly work overtime have a significantly higher risk of attrition
* *BusinessTravel* shows that employees who often have go on business trips tend to quit
* The importance of economic or rational factors is undoubtable and understandable in any case, which is confirmed by the model analysis.
* Moreover, the importnace of features of the current cluster **confirm the second hypothesis** of our research: Employees who frequently travel for work are more likely to experience attrition than those who travel rarely or not at all; as well as **the thrid hypothesis**: Employees who work overtime have a significantly higher risk of leaving the organization compared to employees who do not work overtime.

2. **Employee satisfaction and emotional state**
The models clearly showed that staff turnover is primarily an emotional decision, not just a rational calculation. Most of the strongest predictors of dismissal turned out to be related to how an employee feels at work:
* *EnvironmentSatisfaction* shows how comfortable a person feels in the office ad has quite a significant influence on whether a person is likely to stay in the company
* *JobSatisfaction* reflects satisfaction with tasks, whether they are interesting and meet the needs of the employee, does the employee feel their importance
* *WorkLifeBalance* shows whether a person is satisfied with their recovery after work, or does constant overwork exhaust them
* Practically, this block of fueatures show that employees leave not just for rational reasons, but when they stop enjoying their work or when work begins to destroy their personal lives.
* Moreover, the importnace of features of the current cluster **confirm the first hypothesis** of our research: Employees with low job satisfaction are significantly more likely to leave the organization than employees with higher job satisfaction.

3. **Movement and development in the company**
The last but not least important cluster of factors is related to what happens to an employee's career within the organization:
* *YearsAtCompany* show that long staying in the company increases the probability of attrition. This case potentially arises due to the lack of promotion of employees or decrese in salisfaction levels after several years at the company.
* *YearsSinceLastPromotion* basically shows that if an employee has not received career advancement for a long time, he begins to look for opportunities outside the company
* *NumCompaniesWorked* reflects that employees who have already changed realtively high number of jobs are more likely to do so again
* These results show that stagnation in employees' career paths is one of the main reasons companies cannot hold employees. Even a satisfied employee will start looking for new challenges if he does not see opportunities for growth.
